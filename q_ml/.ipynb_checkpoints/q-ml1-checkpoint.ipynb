```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)

sv = Statevector(qc)        # exact statevector, no backend needed
print(sv)
print(sv.probabilities_dict())
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

sim = AerSimulator()                      # or AerSimulator(method="statevector")
result = sim.run(transpile(qc, sim), shots=1024).result()
print(result.get_counts())
```

```python
from qiskit.circuit.library import ZZFeatureMap
from qiskit_machine_learning.kernels import FidelityQuantumKernel
from qiskit_machine_learning.state_fidelities import ComputeUncompute
from qiskit.primitives import StatevectorSampler

# Feature map
feature_map = ZZFeatureMap(feature_dimension=2, reps=2)

# Sampler
sampler = StatevectorSampler()

# Fidelity calculation
fidelity = ComputeUncompute(sampler=sampler)

# Quantum kernel
quantum_kernel = FidelityQuantumKernel(
    feature_map=feature_map,
    fidelity=fidelity
)

print(quantum_kernel)
```

```python
import numpy as np

# Sample input data
X = np.array([
    [0.1, 0.2],
    [0.4, 0.5],
    [1.0, 1.2],
    [1.5, 1.8]
])

# Calculate kernel matrix
K = quantum_kernel.evaluate(X)

print("Quantum Kernel Matrix:")
print(K)

print("\nMatrix shape:", K.shape)
print("Symmetric:", np.allclose(K, K.T))
```

```python
from qiskit_machine_learning.algorithms import QSVC

y = np.array([0, 0, 1, 1])

qsvc = QSVC(quantum_kernel=quantum_kernel)

qsvc.fit(X, y)

predictions = qsvc.predict(X)

print("Predictions:", predictions)
print("Accuracy:", qsvc.score(X, y))
```
